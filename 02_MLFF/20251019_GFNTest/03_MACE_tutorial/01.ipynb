```python
import numpy as np

# NPZ 파일 경로
path = "../02_XTB_IONPAIR_MD/2pairs/xtbopt/dataset.npz"

# 데이터 로드
data = np.load(path)
print(f"✅ Loaded dataset: {path}")

# 키 확인
print("\n📦 Keys in dataset:")
for key in data.files:
    print(f"  - {key} | shape={data[key].shape}")

# 에너지 분석
if "energy" in data:
    energies = data["energy"]
elif "energies" in data:
    energies = data["energies"]
else:
    energies = None

if energies is not None:
    print("\n⚡ Energy Statistics")
    print(f"  • Number of frames = {len(energies)}")
    print(f"  • Mean energy      = {energies.mean():.6f}")
    print(f"  • Min energy       = {energies.min():.6f}")
    print(f"  • Max energy       = {energies.max():.6f}")
else:
    print("\n⚠️ Energy data not found in dataset")

# 포스 정보
if "forces" in data:
    forces = data["forces"]
    print(f"\n🧲 Force array shape: {forces.shape}  (frames, atoms, 3)")
else:
    print("\n⚠️ Force data not found")
```
