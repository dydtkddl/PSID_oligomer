```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch, Rectangle
import numpy as np

# 한글 폰트 문제 해결
# plt.rcParams['font.family'] = ['DejaVu Sans', 'Arial Unicode MS', 'Malgun Gothic']
# plt.rcParams['axes.unicode_minus'] = False

def create_project_roadmap():
    """전체 26주 프로젝트 로드맵 생성"""
    fig, ax = plt.subplots(figsize=(20, 12))
    ax.set_xlim(0, 28)
    ax.set_ylim(0, 10)
    ax.axis('off')

    # 메인 타이틀
    ax.text(14, 9.5, 'PBP-g-PIL Ion Conductor: Complete Project Roadmap', 
            fontsize=24, weight='bold', ha='center')
    ax.text(14, 9, 'MACE-based MD Simulation for Hopping Mechanism Proof', 
            fontsize=16, style='italic', ha='center', color='gray')

    # Phase 박스들
    phases = [
        ('Phase 1: DFT Dataset (Week 1-8)', 1, 7, 7, 1.8, '#2E86AB', '#A7C6DA', 
         ['CP2K AIMD: 3 temps × 20 ps', 'NEB: Hopping pathways', 'Active Learning: 5 iterations']),
        ('Phase 2: MACE Training (Week 9-12)', 9, 7, 4, 1.8, '#F18F01', '#FDCB9E',
         ['MACE-OFF-μ Fine-tuning', '15,000 structures', 'E/F/Dipole multi-task']),
        ('Phase 3: Production MD (Week 13-20)', 14, 7, 7, 1.8, '#C73E1D', '#F4A6A3',
         ['LAMMPS + MACE: 4 μs total', 'Target + 3 Control systems', 'PLUMED Metadynamics']),
        ('Phase 4: Analysis (Week 21-26)', 22, 7, 4, 1.8, '#6A4C93', '#B8A6D3',
         ['Van Hove Function', 'Anisotropic Diffusion', 'Hopping Detection'])
    ]

    for title, x, y, w, h, edge_color, face_color, items in phases:
        # Phase 박스
        phase_box = FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.1", 
                                  edgecolor=edge_color, facecolor=face_color, linewidth=3)
        ax.add_patch(phase_box)
        
        # Phase 제목
        ax.text(x + w/2, y + h - 0.3, title, fontsize=14, weight='bold', ha='center')
        
        # Phase 내용
        for i, item in enumerate(items):
            ax.text(x + w/2, y + h - 0.7 - i*0.3, f'• {item}', fontsize=10, ha='center')

    # 연결 화살표
    arrows = [(8, 8, 9, 8), (13, 8, 14, 8), (21, 8, 22, 8)]
    for x1, y1, x2, y2 in arrows:
        arrow = FancyArrowPatch((x1, y1), (x2, y2), arrowstyle='->', 
                               mutation_scale=30, linewidth=3, color='black')
        ax.add_patch(arrow)

    # 도구 스택
    ax.text(14, 6.2, 'Computational Tool Stack', fontsize=16, weight='bold', ha='center')
    
    tool_categories = [
        ('Structure', ['CREST', 'Avogadro', 'Packmol'], '#2E86AB'),
        ('DFT/QM', ['CP2K', 'xTB', 'JDFTx'], '#F18F01'),
        ('MLIP', ['MACE-OFF-μ', 'ASE'], '#C73E1D'),
        ('MD', ['LAMMPS', 'OpenMM'], '#6A4C93'),
        ('Analysis', ['PLUMED', 'MDAnalysis'], '#0B6E4F'),
        ('Viz', ['VMD', 'PyMOL', 'Jmol'], '#9D4EDD')
    ]

    x_start = 2
    for cat_name, tools, color in tool_categories:
        cat_box = FancyBboxPatch((x_start, 5.2), 3.5, 0.6, 
                                boxstyle="round,pad=0.05",
                                edgecolor=color, facecolor='white', linewidth=2)
        ax.add_patch(cat_box)
        ax.text(x_start+1.75, 5.5, cat_name, fontsize=11, weight='bold', ha='center')
        ax.text(x_start+1.75, 4.9, ', '.join(tools), fontsize=8, ha='center', style='italic')
        x_start += 4.2

    # 핵심 성과물
    ax.text(14, 3.8, 'Key Scientific Deliverables', fontsize=16, weight='bold', ha='center')
    
    deliverables = [
        ('Mechanism Proof', 'Van Hove peaks\nHopping events > 100', 3),
        ('Quantitative Match', 'σ_sim vs σ_exp\n±20% accuracy', 8),
        ('Free Energy', 'ΔF‡ < 0.25 eV\nπ-π stacking effect', 13),
        ('Publication', 'ACS Appl. Mater.\nInterfaces Target', 18),
        ('Design Rules', 'Optimal composition\nPredictive guidelines', 23)
    ]

    for title, desc, x_pos in deliverables:
        deliv_box = FancyBboxPatch((x_pos-1.5, 1.5), 3, 1.8,
                                  boxstyle="round,pad=0.1",
                                  edgecolor='#0B6E4F', facecolor='#C7F0DB', linewidth=2)
        ax.add_patch(deliv_box)
        ax.text(x_pos, 2.8, title, fontsize=10, weight='bold', ha='center')
        ax.text(x_pos, 2.2, desc, fontsize=8, ha='center', va='center')

    plt.tight_layout()
    plt.savefig('project_roadmap.png', dpi=300, bbox_inches='tight', facecolor='white')
    return fig

def create_voltage_strategy():
    """전압 시뮬레이션 전략 다이어그램"""
    fig, ax = plt.subplots(figsize=(16, 10))
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 10)
    ax.axis('off')

    # 타이틀
    ax.text(5, 9.5, 'Electric Field Simulation Strategy', fontsize=22, weight='bold', ha='center')
    ax.text(5, 9, 'Equilibrium vs Non-equilibrium MD Approaches', fontsize=14, style='italic', ha='center')

    # Phase 1: 평형 (필수)
    eq_rect = Rectangle((0.5, 5.5), 4, 3, edgecolor='#2E86AB', 
                       facecolor='#A7C6DA', linewidth=3)
    ax.add_patch(eq_rect)
    ax.text(2.5, 8.2, 'Phase 1: Equilibrium MD', fontsize=14, weight='bold', ha='center')
    ax.text(2.5, 7.8, '(Week 1-24) - ESSENTIAL', fontsize=11, ha='center', style='italic', color='red')

    eq_items = [
        'Zero-bias conductivity σ(0V)',
        'Einstein relation: D → σ',
        'PLUMED: Hopping free energy',
        'Direct experimental comparison',
        'Nernst-Einstein validation'
    ]
    y_pos = 7.3
    for item in eq_items:
        ax.text(2.5, y_pos, f'• {item}', fontsize=9, ha='center')
        y_pos -= 0.3

    ax.text(2.5, 5.8, '✓ Scientific Rigor', fontsize=10, weight='bold', 
           ha='center', color='green')

    # Phase 2: 비평형 (확장)
    nemd_rect = Rectangle((5.5, 5.5), 4, 3, edgecolor='#C73E1D',
                         facecolor='#F4A6A3', linewidth=3)
    ax.add_patch(nemd_rect)
    ax.text(7.5, 8.2, 'Phase 2: NEMD (Extension)', fontsize=14, weight='bold', ha='center')
    ax.text(7.5, 7.8, '(Week 25-50) - HIGH VALUE', fontsize=11, ha='center', style='italic', color='orange')

    nemd_items = [
        'Applied E-field: 0.01-0.1 V/Å',
        'Directed transport visualization',
        'Anisotropic response analysis',
        'Hopping pathway movies',
        'Target vs Control comparison'
    ]
    y_pos = 7.3
    for item in nemd_items:
        ax.text(7.5, y_pos, f'• {item}', fontsize=9, ha='center')
        y_pos -= 0.3

    ax.text(7.5, 5.8, '✓ Visual Impact', fontsize=10, weight='bold',
           ha='center', color='orange')

    # 연결 화살표
    arrow = FancyArrowPatch((4.5, 7), (5.5, 7), arrowstyle='->', 
                           mutation_scale=30, linewidth=3, color='black')
    ax.add_patch(arrow)
    ax.text(5, 7.3, 'Extension', fontsize=10, ha='center', weight='bold')

    # 기술적 고려사항
    tech_box = FancyBboxPatch((1, 0.5), 8, 4.5, boxstyle="round,pad=0.15",
                             edgecolor='#6A4C93', facecolor='#E8DFF5', linewidth=2)
    ax.add_patch(tech_box)
    ax.text(5, 4.6, 'Technical Implementation Strategy', fontsize=14, weight='bold', ha='center')

    # 3컬럼 구성
    considerations = [
        ('Equilibrium Approach', [
            'Advantages:',
            '• Direct exp. comparison',
            '• Thermodynamically exact',
            '• MACE fully validated',
            '',
            'Einstein Relation:',
            'σ = (ne²D)/(k_BT)',
            'D from MSD analysis'
        ], 2),
        ('NEMD Implementation', [
            'LAMMPS Command:',
            'fix efield tfsi efield 0.1 0 0',
            '',
            'Scale Challenge:',
            '• Realistic: ~1e-6 V/Å',
            '• Observable: ~0.1 V/Å',
            '• Gap: 10⁵ factor',
            'Use for mechanism viz'
        ], 5),
        ('Strategic Priority', [
            'Phase 1 (Must-have):',
            '✓ Mechanism proof',
            '✓ Quantitative prediction',
            '✓ Publication ready',
            '',
            'Phase 2 (Nice-to-have):',
            '□ Visual demonstration',
            '□ Device connection',
            '□ Extended impact'
        ], 8)
    ]

    for title, items, x_pos in considerations:
        ax.text(x_pos, 4.2, title, fontsize=11, weight='bold', ha='center',
               bbox=dict(boxstyle='round', facecolor='white', edgecolor='black'))
        y_pos = 3.8
        for item in items:
            if item.startswith(('•', '□', '✓')):
                ax.text(x_pos, y_pos, item, fontsize=8, ha='center')
            elif item.endswith(':'):
                ax.text(x_pos, y_pos, item, fontsize=9, ha='center', weight='bold')
            else:
                ax.text(x_pos, y_pos, item, fontsize=8, ha='center', style='italic')
            y_pos -= 0.25

    plt.tight_layout()
    plt.savefig('voltage_strategy.png', dpi=300, bbox_inches='tight', facecolor='white')
    return fig

def create_plumed_mechanism():
    """PLUMED 메커니즘 설명 다이어그램"""
    fig, ax = plt.subplots(figsize=(16, 10))
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 10)
    ax.axis('off')

    # 타이틀
    ax.text(5, 9.5, 'PLUMED Enhanced Sampling: Solving the Rare Event Problem', 
           fontsize=20, weight='bold', ha='center')

    # Problem vs Solution
    problem_box = FancyBboxPatch((0.5, 6.5), 4, 2.5, boxstyle="round,pad=0.15",
                                edgecolor='red', facecolor='#FFE5E5', linewidth=3)
    ax.add_patch(problem_box)
    ax.text(2.5, 8.7, '❌ Problem: Rare Event', fontsize=14, weight='bold', ha='center', color='red')
    ax.text(2.5, 8.2, 'Normal MD (100 ns):', fontsize=11, ha='center')
    ax.text(2.5, 7.8, 'TFSI⁻ hopping: 0-3 events', fontsize=10, ha='center')
    ax.text(2.5, 7.4, 'Energy barrier: 0.2-0.3 eV', fontsize=10, ha='center')
    ax.text(2.5, 7.0, 'Time scale mismatch:', fontsize=10, ha='center')
    ax.text(2.5, 6.6, 'MD: ns | Hopping: μs-ms', fontsize=9, ha='center', style='italic')

    solution_box = FancyBboxPatch((5.5, 6.5), 4, 2.5, boxstyle="round,pad=0.15",
                                 edgecolor='green', facecolor='#E5FFE5', linewidth=3)
    ax.add_patch(solution_box)
    ax.text(7.5, 8.7, '✅ Solution: PLUMED', fontsize=14, weight='bold', ha='center', color='green')
    ax.text(7.5, 8.2, 'Metadynamics (50 ns):', fontsize=11, ha='center')
    ax.text(7.5, 7.8, 'TFSI⁻ hopping: 100+ events', fontsize=10, ha='center')
    ax.text(7.5, 7.4, 'Free energy: ΔF‡ ± 0.01 eV', fontsize=10, ha='center')
    ax.text(7.5, 7.0, 'Complete landscape mapping:', fontsize=10, ha='center')
    ax.text(7.5, 6.6, 'All pathways explored', fontsize=9, ha='center', style='italic')

    # 큰 화살표
    big_arrow = FancyArrowPatch((4.5, 7.75), (5.5, 7.75), arrowstyle='->', 
                               mutation_scale=50, linewidth=5, color='blue')
    ax.add_patch(big_arrow)

    # 메커니즘 설명
    mech_box = FancyBboxPatch((1, 3.5), 8, 2.5, boxstyle="round,pad=0.15",
                             edgecolor='#2E86AB', facecolor='#D6EAF8', linewidth=2)
    ax.add_patch(mech_box)
    ax.text(5, 5.7, 'How PLUMED Works: Gaussian Hill Metadynamics', 
           fontsize=14, weight='bold', ha='center')

    # 3단계 프로세스
    steps = [
        ('1️⃣', 'Define Collective Variable\nCoordination: TFSI-VBBI', 2),
        ('2️⃣', 'Add Gaussian Bias\nEvery 500 steps\nPush to new states', 5),
        ('3️⃣', 'Well-Tempered Convergence\nHills decay automatically\nFree energy reconstructed', 8)
    ]

    for emoji, desc, x_pos in steps:
        ax.text(x_pos, 5.0, emoji, fontsize=20, ha='center')
        ax.text(x_pos, 4.3, desc, fontsize=9, ha='center', va='center')

    # 핵심 수식
    formula_box = FancyBboxPatch((1, 0.5), 8, 2.5, boxstyle="round,pad=0.15",
                                edgecolor='#6A4C93', facecolor='#F4ECF7', linewidth=2)
    ax.add_patch(formula_box)
    ax.text(5, 2.7, 'Mathematical Framework', fontsize=14, weight='bold', ha='center')

    formulas = [
        'CV = Σ [1-(r/r₀)⁶] / [1-(r/r₀)¹²]  (Coordination Number)',
        'V_bias(t) = Σ w·exp[-(CV-CV_i)²/2σ²]  (Gaussian Hills)',
        'F(CV) = -lim(t→∞) V_bias(CV,t)  (Free Energy Surface)',
        'w_eff = w₀·exp[-V_bias/(k_B T·Δ)]  (Well-Tempered Decay)'
    ]

    y_pos = 2.2
    for formula in formulas:
        ax.text(5, y_pos, formula, fontsize=9, ha='center', family='monospace',
               bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))
        y_pos -= 0.35

    plt.tight_layout()
    plt.savefig('plumed_mechanism.png', dpi=300, bbox_inches='tight', facecolor='white')
    return fig

# 모든 다이어그램 생성 실행
if __name__ == "__main__":
    print("🎨 Generating strategic diagrams for PBP-g-PIL project...")
    
    fig1 = create_project_roadmap()
    print("✅ project_roadmap.png - Complete 26-week roadmap")
    
    fig2 = create_voltage_strategy()
    print("✅ voltage_strategy.png - Electric field simulation strategy")
    
    fig3 = create_plumed_mechanism()
    print("✅ plumed_mechanism.png - Enhanced sampling explanation")
    
    plt.show()
    print("\n🎯 All diagrams ready for 부경대 meeting!")
    print("📁 Files saved: project_roadmap.png, voltage_strategy.png, plumed_mechanism.png")
```
